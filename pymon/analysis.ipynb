# PML File Analysis Notebook

Interactive exploration of Process Monitor log files using procmon-parser

```python
from analyze_pml import PMLAnalyzer
import pandas as pd
from collections import Counter

# Change this to your PML file path
pml_file = "your_file.pml"

print(f"Loading {pml_file}...")
analyzer = PMLAnalyzer(pml_file)
print(f"Loaded {len(analyzer.events)} events")
```

## Summary Statistics

```python
analyzer.print_summary()
```

## Convert to DataFrame for Analysis

```python
df = pd.DataFrame(analyzer.events)
print(f"DataFrame shape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
df.head()
```

## File Operations Analysis

```python
file_ops = analyzer.get_file_operations()
print(f"Total file operations: {len(file_ops)}")

file_df = pd.DataFrame(file_ops)
print(f"\nTop 10 files accessed:")
file_df['Path'].value_counts().head(10)
```

## Registry Operations Analysis

```python
reg_ops = analyzer.get_registry_operations()
print(f"Total registry operations: {len(reg_ops)}")

if reg_ops:
    reg_df = pd.DataFrame(reg_ops)
    print(f"\nTop 10 registry keys:")
    reg_df['Path'].value_counts().head(10)
```

## Process Analysis

```python
# Select a specific process to analyze
process_name = "explorer"  # Change this

process_events = analyzer.get_events_by_process(process_name)
print(f"Events for {process_name}: {len(process_events)}")

if process_events:
    proc_df = pd.DataFrame(process_events)
    print(f"\nTop operations by {process_name}:")
    proc_df['Operation'].value_counts().head(10)
```

## Export Analysis

```python
# Export file operations to CSV
analyzer.export_to_csv("file_operations.csv", file_ops)

# Export registry operations to CSV
analyzer.export_to_csv("registry_operations.csv", reg_ops)
```
